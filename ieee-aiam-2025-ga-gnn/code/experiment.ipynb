# Quantum-Inspired Hybrid Genetic Algorithm and Graph Neural Network Ensemble for Multimodal Classification

Install the notebook extra in the kernel environment. Set paths to your table, modality configuration and a new or empty output directory in the next cell.

```python
import json
from pathlib import Path

import pandas as pd

from aiam_gnn.cli import main

input_file = Path("measurements.csv")
modalities = Path("modalities.json")
label = "target"
positive_class = "1"
encoder = "gcn"
output = Path("notebook_results")
```

```python
arguments = [
    "--input",
    str(input_file),
    "--modalities",
    str(modalities),
    "--label",
    label,
    "--positive",
    positive_class,
    "--encoder",
    encoder,
    "--seed",
    "42",
    "--output",
    str(output),
    "--explain",
]
main(arguments)
```

```python
metrics = json.loads((output / "metrics.json").read_text(encoding="utf-8"))
pd.DataFrame.from_dict(metrics, orient="index")
```
